# Análisis estadístico OT
Calibración 2025; resultados posteriores no son entradas de programación.

```
from modelo import datos, ajustes
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
p,d,c,cal,e=datos()
print(p.shape, d.shape, c.shape, cal.shape, e.shape)
print(p.isna().sum())
```

```
for col, rows in ajustes().items():
    display(pd.DataFrame(rows))
    x=d.loc[(d.anio_programacion==2025)&(d.ejecutada==1),col].dropna()
    x=x[x>0]
    best=rows[0]; f=getattr(stats,best["familia"]); par=best["parametros"]
    grid=np.linspace(0,x.quantile(.995),300)
    plt.hist(x,bins=60,density=True); plt.plot(grid,f.pdf(grid,*par)); plt.title(col); plt.show()
    q=np.linspace(.01,.99,99); plt.scatter(f.ppf(q,*par),np.quantile(x,q)); plt.xlabel("Cuantiles teóricos"); plt.ylabel("Cuantiles observados"); plt.title("Q-Q: "+col); plt.show()
```
